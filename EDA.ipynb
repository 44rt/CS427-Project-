# Exploratory Data Analysis

This is to have an overview of the datasets such as viewing the size of the datasets, datatypes, null values, comparison graphs of how much benign traffic there is to malicious traffic as well as a graph showing the types of attacks that occur in the datasets.

```python
# Importing libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# make plots look nicer
sns.set(style="whitegrid")
```

```python
# Load raw CSVs
df1 = pd.read_csv('/Users/roskat/Desktop/CS427/data/BTS_1.csv')
df2 = pd.read_csv('/Users/roskat/Desktop/CS427/data/BTS_2.csv')

# Shape of datasets - number of rows and columns
print("BTS1 shape:", df1.shape)
print("BTS2 shape:", df2.shape)
```

# BTS1 (Base Transceiver Station 1) Dataset Overview

```python
# Quick overview & missing values
display(df1.head())
display(df1.info())
display(df1.describe(include='all'))

print("Missing values (BTS1):")
display(df1.isna().sum().sort_values(ascending=False).head(20))
```

```python
# BTS1 - Label distribution and attack types
plt.figure(figsize=(6,4))
df1['Label'].value_counts().plot(kind="bar")
plt.title("Benign vs Malicious (BTS1)")
plt.show()

# Attack-type frequencies (robust check of possible column names)
for attack_col in ['Attack Type', 'AttackType', 'Attack_Type', 'Attack']:
    if attack_col in df1.columns:
        plt.figure(figsize=(10,4))
        df1[attack_col].value_counts().plot(kind="bar")
        plt.title(f"Attack Types (BTS1) - column: {attack_col}")
        plt.show()
        break
```

```python
# BTS1 - Correlations (numeric features)
num_cols1 = df1.select_dtypes(include='number')
plt.figure(figsize=(12,8))
sns.heatmap(num_cols1.corr(), cmap="coolwarm", center=0)
plt.title("Numeric Feature Correlations (BTS1)")
plt.show()
```
